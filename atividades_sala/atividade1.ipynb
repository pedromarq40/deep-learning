```python
import numpy as np

def sigmoid(z : float) -> float:
    return 1 / ( 1 + np.exp(-1 * z))
```

```python
from numpy.typing import NDArray

def neuron_output(x : NDArray, w : NDArray, b : float) -> float:

    produto = x.dot(w.T)
    print(f'Produto das entradas com os pesos: {produto}')
    z = produto + b
    print(f'Soma com o vies: {z}')
    return sigmoid(z)
```

```python
x = np.array([0.5, -0.3, 0.8])
w = np.array([1.2, 0.7, -0.5])
b = -0.2

a = neuron_output(x, w, b)
print(f'Ativação {a}')
```
