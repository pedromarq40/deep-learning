```python
import numpy as np 

def softmax(z):
    max = np.max(z)
    return np.exp(z - max) / np.sum(np.exp(z - max))

def cross_entropy(y_true, y_pred):
    j = -np.sum(y_true * np.log(y_pred))
    return j
```

```python
z = np.array([1.5, 0.5, -0.5])
y = np.array([1, 0, 0])
y_pred = softmax(z)
j = cross_entropy(y, y_pred)

print(f'Após softmax: {y_pred}')
print(f'Perda: {j}')
```
