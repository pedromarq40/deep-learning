```python
import numpy as np

np.random.seed(0)

def sigmoid(z : float) -> float:
    return 1 / ( 1 + np.exp(-1 * z))

print(sigmoid(-0.21))
```

```python
from numpy.typing import NDArray

def neuron_output(x : NDArray, w : NDArray, b : float) -> float:

    produto = x.dot(w)
    z = produto + b
    return sigmoid(z)
```

```python
x = np.array([0.5, -0.3, 0.8])
w = np.array([1.2, 0.7, -0.5])
b = -0.2

a = neuron_output(x, w, b)
print(f'Ativação {a}')
```

```python
def softmax(z):
    maior_linha = np.max(z, axis=1, keepdims=True)
    exp = np.exp(z - maior_linha)
    soma_linha = np.sum(exp, axis=1, keepdims=True)
    return exp / soma_linha

def feed_forward(X, W1, b1, W2, b2, W3, b3):
    print(f'Dimensão da entrada: {X.shape}')
    print(f'Dimensão da matriz W1 de pesos: {W1.shape}')
    z2 = X.dot(W1) + b1
    a2 = sigmoid(z2)
    print(f'Dimensão da primeira camada oculta: {a2.shape}')
    print(f'Dimensão da matriz W2 de pesos: {W2.shape}')
    z3 = a2.dot(W2) + b2
    a3 = sigmoid(z3)
    print(f'Dimensão da segunda camada oculta: {a3.shape}')
    print(f'Dimensão da matriz W3 de pesos: {W3.shape}')
    z4 = a3.dot(W3) + b3
    a4 = softmax(z4)
    print(f'Dimensão da saída | softmax: {a4.shape}')
    return a4

X = np.random.rand(300, 3)
W1 = np.random.rand(3, 4)
b1 = np.random.rand(1, 1)
W2 = np.random.rand(4, 4)
b2 = np.random.rand(1, 1)
W3 = np.random.rand(4, 3)
b3 = np.random.rand(1, 1)

y_ = feed_forward(X, W1, b1, W2, b2, W3, b3)

print(np.sum(y_, axis=1))
```
